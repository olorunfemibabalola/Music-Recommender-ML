````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
````

Import the Data

````python
music_data = pd.read_excel('/content/music copy.xlsx')
music_data
````

````python
music_data.describe()
````

````python
music_data.shape
````

````python
music_data.genre.dtype
````

Clean the Data

````python
#The data is already clean
````

Split the Data into Training/ Test Sets

````python
X = music_data.drop(columns=["genre"])
y = music_data["genre"]

````

Create a Model

````python
model = DecisionTreeClassifier()
````

Train the Model

````python
model.fit(X, y)
````

Make Predictions

````python
age = None
while age is None:
    try:
        age_input = input("Enter the age of the person you want to predict: \n")
        age = int(age_input)
        if age < 0:
            print("Age cannot be negative. Please try again.")
            age = None
    except ValueError:
        print("Invalid input. Please enter a valid integer for the age.")
        age = None

gender = None
while gender is None:
    gender_input = input("Enter the gender of the person you want to predict (male/female): \n").lower().strip()
    if gender_input == "male":
        gender = 1
    elif gender_input == "female":
        gender = 0
    else:
        print("Invalid input. Please enter 'male' or 'female'.")

try:
    predictions = model.predict(
        [[age, gender]]
    )
    predictions = str(predictions)
    predictions = predictions.strip("[]")
    predictions = predictions.strip("''")
    predictions = predictions.lower()

    print(f'\nYour model predicts that the type of music this person would listen to would be {predictions}\n\n')
except Exception as e:
    print(f"An error occurred during prediction: {e}")
````

Evaluate and Improve

````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = model.fit(X_train, y_train)

predictions = model.predict(X_test)

score = accuracy_score(y_test, predictions)

score = float(score * 100)

print(f"Your model's accuracy is {score}%")

````
