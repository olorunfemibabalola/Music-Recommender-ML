# Music Preference Predictor

- Employs a simple and interpretable Decision Tree model.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
````

Import the Data

````python
music_data = pd.read_excel('/content/music copy.xlsx')
music_data
````

````python
music_data.describe()
````

````python
music_data.shape
````

````python
music_data.genre.dtype
````

Clean the Data

````python
#The data is already clean
````

Split the Data into Training/ Test Sets

````python
X = music_data.drop(columns=["genre"])
y = music_data["genre"]

````

Create a Model

````python
model = DecisionTreeClassifier()
````

Train the Model

````python
model.fit(X, y)
````

Make Predictions

````python
while True:
  try:
      age = input("Enter the age of the person you want to predict: \n")
      age = int(age)
  except ValueError:
      print("Invalid input. Please enter a valid integer for the age.")
      exit()

  if age < 0:
      print("Age cannot be negative.")
      exit()

  try:
    gender = input("Enter the gender of the person you want to predict: \n")
    if gender.lower().strip() == "male":
      gender = 1
    elif gender.lower().strip() == "female":
      gender = 0
      break
    else:
      print("Invalid input. Please enter 'male' or 'female'.")
      exit()
  except:
    print("Invalid input. Please enter 'male' or 'female'.")
    exit()

predictions = model.predict([
                [age, gender]
                ])

print(f'\nYour model predicts: {predictions}\n\n')
````
